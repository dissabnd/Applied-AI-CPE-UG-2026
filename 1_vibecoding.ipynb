### The kernel remembers

Behind the notebook sits a **kernel** — a Python session that stays alive between cells. Anything
you define in one cell is available in every later cell.

Run the next two cells in order.


```python
reflux_ratio = 1.8          # defined here...
feed_temp = 92.0
```

```python
# ...and still available here, in a different cell
print(f"reflux ratio = {reflux_ratio}, feed temperature = {feed_temp} °C")
```

### Execution order is *your* order, not top-to-bottom

The number in `In [ ]:` to the left of a code cell tells you **when** it ran, not where it sits.
A notebook that runs perfectly for you can fail for someone else if you ran the cells out of order.

Run the cell below **three times in a row** and watch the answer change.


```python
batch_count = 0     # <- this line only runs when this whole cell runs

batch_count = batch_count + 1
print("batches:", batch_count)
```

Now delete the first line and run it three more times — you will get an error the first time and
a growing number after that. This is the single most common source of "but it worked yesterday".

**The habit that saves you:** before you share a notebook or trust a result, do
**Kernel → Restart & Run All**. If it does not survive that, it is not finished.

## 1 · Vibe coding

**Vibe coding** is writing software by describing what you want in plain language and letting a
large language model produce the code. You stay the engineer: you set the goal, judge the output,
and decide what to keep. The model is fast and fluent, and confidently wrong often enough to matter.

It is very good for: exploring a dataset, drafting a plot, trying an idea in ten minutes instead of
an afternoon.

It is risky for: anything you will not read carefully afterwards.

**Tools you can use today**

- Any chat LLM — ChatGPT, Claude, Gemini, Grok, DeepSeek
- Google Colab (Gemini built in) — nothing to install
- Cursor or Windsurf — an editor with the model inside it


### Writing a prompt that works

A weak prompt gets you plausible code for the wrong problem. Five things make the difference:

1. **Context** — who you are and what the data is. *"I have a pandas DataFrame of cement kiln data."*
2. **Task** — one clear job. *"Fit a linear regression predicting CO₂ emissions."*
3. **Data shape** — column names, units, number of rows. Paste `df.head()` if you have it.
4. **Output** — what you want back. *"Print the equation, R² and RMSE."*
5. **Constraints** — libraries, style. *"Use scikit-learn and matplotlib. No seaborn."*

Two habits that pay off immediately:

- **One step at a time.** Ask for the plot, run it, then ask for the model. A 200-line first answer
  is a 200-line debugging job.
- **Ask it to explain.** *"Explain what this code does, line by line."* If the explanation does not
  match what you asked for, the code will not either.


### Then check it — every time

Generated code that runs is not generated code that is right. Before you use a result:

- [ ] Does it run from a **restarted kernel**, top to bottom?
- [ ] Did it use **my** data, or did it quietly invent its own?
- [ ] Are the **units** right? Are the axis labels right?
- [ ] Is the magnitude **physically sensible**? A negative efficiency, a 400 °C feed, an R² of exactly 1.0 — all suspicious.
- [ ] Can I say **in one sentence** what every line does?
- [ ] Would I put this number in a report with my name on it?

> The last box is the whole point of this module. The model is responsible for none of this. You are.
